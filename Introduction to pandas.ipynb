Controlling axes using xlim and ylim
and diagonals

```python
g=sns.FacetGrid(movies, row= 'Genre', col= 'Year', hue='Genre')
kws= dict(s= 50, linewidth= 0.5, edgecolor= 'blue')
g= g.map(plt.scatter, 'CriticalRating', 'AudienceRatings',**kws)
g.set(xlim= (0,100), ylim=(0,100))
plt.show()
```

```python
g=sns.FacetGrid(movies, row= 'Genre', col= 'Year', hue='Genre')
kws= dict(s= 50, linewidth= 0.5, edgecolor= 'blue')
g= g.map(plt.scatter, 'CriticalRating', 'AudienceRatings',**kws)
g.set(xlim= (0,100), ylim=(0,100))
for ax in g.axes.flat:
    ax.plot((0,100), (0,100), ls='--', c='gray')
    g.add_legend()

plt.show()
```
