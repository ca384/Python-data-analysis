### To draw a christmas tree

1. ask the user to type in the tree size
2. print the top of the tree
3. print the bottom of the tree

```python
#top of the tree
print('Enter the tree size:')
size = int(input())
for row_num in range(1, size + 1):
    space= ' ' * (size - row_num) #* means repeat
    tree = '^' * (row_num * 2 -1)
    print(space + tree)

 #bottom of the tree
spaces = ' ' * (size - 1)
print(spaces  + '#')
print(spaces + '#')


```

#### use a while loop to make the christmas tree

```python
print('Enter the tree size: ')
size = int(input())
row_num = 1
while row_num < size + 1:
    spaces = ' ' * (size - row_num)
    tree = '^' * (row_num * 2-1)
    print(spaces + tree)
    row_num = row_num + 1

# to print the trunk of the tree
spaces = ' ' * (size - 1)
print(spaces + '#')
print(spaces + '#')

```

### To make a christmas tree

```python
size = int(input('Enter the size of the tree: '))
for row in range(1, size + 1):
    space = size - row
    star = 2 * row - 1
    print(' ' * space +'^' * star)
for _ in range(2):
    print(' ' * (size-1) + '#')
```
