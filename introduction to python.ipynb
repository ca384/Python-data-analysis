```python
from numpy.random import randn
```

```python
randn()
```

```python
x = randn()
if x > 1:
    answer = 'Greater than 1'
print(x)
print(answer)
```

```python
answer = None
x = randn()
if x > 1:
    answer = 'Greater than 1'
print(x)
print(answer)
```

```python
x = randn()
if x > 1:
    print('Greater than 1')
print(x)
```

```python
answer = None
x = randn()
if x > 1:
    answer = 'Greater than 1'
else:
    answer = 'Less than 1'
print(x)
print(answer)
```

```python
answer = None
x = randn()
if x > 1:
    answer = 'Greater than 1'
else:
    if x >= -1:
        answer = 'Between -1 and 1'
    else:
        answer = 'Less than 1'
print(x)
print(answer)
```
