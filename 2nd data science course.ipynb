### print random number that fall betteen a section of the normal distribution table

```python
import random

import matplotlib_inline
import numpy as np
from numpy.random import randn
```

```python
N=10
counter = 0
for i in randn(N):
    if i <= 1 and i >= 0:
        counter +=1
answer = counter/N
print(answer)
```

We want to be able to input more N

```python
while True:
    N = int(input('Enter N '))
    if N == 0:
        break

    counter = 0

    for i in randn(N):
        if 0 <= i <=1:
            counter = counter + 1
    answer = counter/N
    print('N =', N)
    print('Estimated value =', answer)

```
